```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

uploaded = files.upload()

file_name = next(iter(uploaded))

print("File uploaded successfully!")
```

```python
original_df = pd.read_excel(
    file_name,
    sheet_name="Original_Dataset"
)

df = pd.read_excel(
    file_name,
    sheet_name="Cleaned_Dataset"
)

df["Date"] = pd.to_datetime(df["Date"])

print("Original Dataset:")
display(original_df.head())

print("Cleaned Dataset:")
display(df.head())

print("Original Shape:", original_df.shape)
print("Cleaned Shape:", df.shape)
```

```python
print("Missing Values Before Cleaning:")
print(original_df.isnull().sum())

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    df["Date"],
    df["Turbidity_NTU"],
    marker="o",
    label="Turbidity"
)

plt.title("Water Turbidity Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Turbidity (NTU)")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    df["Date"],
    df["Nitrate_mg_L"],
    marker="o"
)

plt.title("Nitrate Pollution Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Nitrate (mg/L)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    df["Date"],
    df["Lead_mg_L"],
    marker="o"
)

plt.title("Lead Concentration Trend")
plt.xlabel("Date")
plt.ylabel("Lead (mg/L)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
location_avg = df.groupby("Location")[
    ["Turbidity_NTU", "Nitrate_mg_L", "Lead_mg_L"]
].mean()

location_avg.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Water Pollution Indicators by Location")
plt.xlabel("Location")
plt.ylabel("Average Value")
plt.xticks(rotation=30)
plt.legend(title="Parameters")
plt.grid(axis="y")
plt.tight_layout()
plt.show()
```

```python
columns = [
    "pH",
    "Dissolved_Oxygen_mg_L",
    "Turbidity_NTU",
    "TDS_mg_L",
    "Nitrate_mg_L",
    "Lead_mg_L",
    "Temperature_C"
]

correlation = df[columns].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Between Water Quality Parameters")
plt.tight_layout()
plt.show()
```
